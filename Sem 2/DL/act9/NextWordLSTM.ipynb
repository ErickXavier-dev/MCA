```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Flatten, Embedding, Dropout
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.selections import train_test_split
```

```python
corpus = [
    "I love deep learning",
    "Deep learning is amazing",
    "I enjoy machine learning",
    "Machine learning is fascinating",
    "I am passionate about AI",
    "AI is the future",
    "I want to learn more about deep learning",
    "Deep learning is a subset of machine learning",
    "AI is transforming the world",
    "I am excited about the possibilities of AI"
]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
total_words = len(tokenizer.word_index) + 1

input_sequences = []
for line in corpus:
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```
