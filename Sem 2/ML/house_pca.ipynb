```python
data = pd.read_csv('/content/drive/MyDrive/ML_csv/kc_house_data.csv')
data.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data.drop(['id', 'date', 'price'], axis=1))
```

```python
scaled_data
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(scaled_data)
```

```python
pca.fit(scaled_data)
```

```python
x_pca = pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
x_pca.shape
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x_pca[:, 0], x_pca[:, 1], alpha=0.5)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of House Data')
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x_pca[:, 0], x_pca[:, 1], c=x_pca[:, 0], cmap='viridis', alpha=0.5)
plt.colorbar(label='Principal Component 1')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of House Data with Color Gradient')
plt.show()
```
