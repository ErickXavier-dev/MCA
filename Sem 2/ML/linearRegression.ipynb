```python
X = df['Head Size(cm^3)'].values
Y = df['Brain Weight(grams)'].values
```

```python
np.corrcoef(X,Y)
```

```python
plt.scatter(X, Y, c = 'green', label = 'Scatter Plot')
plt.xlabel('Head Size in cm^3')
plt.ylabel('Brain Weight in grams')
plt.legend()
plt.show()
```

```python
mean_x = np.mean(X)
mean_y = np.mean(Y)

n = len(X)

numer = 0
denom = 0

for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] - mean_y)
    denom += (X[i] - mean_x) ** 2
b1 = numer / denom
b0 = mean_y - (b1 * mean_x)

print("Coefficients for regression line: ", b0, b1)
```
